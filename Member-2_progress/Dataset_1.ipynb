```python
print(df['Dominant_Polymer'].unique())
print(df['Dominant_Morphology'].unique())
print(df['Risk_Level'].unique())
```

```python
df['Risk_Level'] = df['Risk_Level'].str.strip().str.title()
df['Dominant_Polymer'] = df['Dominant_Polymer'].str.strip().str.upper()
df['Dominant_Morphology'] = df['Dominant_Morphology'].str.strip().str.title()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:")
print(df.isnull().sum())

print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Sample IDs:", df['Sample_ID'].duplicated().sum())
```
