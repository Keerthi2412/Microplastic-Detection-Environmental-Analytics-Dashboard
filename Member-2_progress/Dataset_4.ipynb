```python
df['Model_Confidence_Pct'] = df['Model_Confidence_Score'] * 100
```

```python
df['Anomaly_Flag'].value_counts()
```

```python
df['Prediction_Error'] = (
    df['Predicted_Microplastic_Count'] -
    df['Actual_Microplastic_Count']
)
```

```python
df['Absolute_Error'] = abs(df['Prediction_Error'])
```

```python
df['Error_Percentage'] = (
    df['Absolute_Error'] /
    df['Actual_Microplastic_Count'].replace(0, 1)
) * 100
```

```python
df[df['Error_Percentage'] > 20]
```
